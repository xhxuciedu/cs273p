```python
import torch
import numpy as np
import torch.nn as nn

def get_memory_usage(tensor):
  """Calculates the memory usage of a PyTorch tensor in bytes.

  Args:
    tensor: The PyTorch tensor.

  Returns:
    The memory usage of the tensor in bytes.
  """
  return tensor.element_size() * tensor.nelement()

x = torch.tensor([[1., 2, 3], [4, 5, 6]])  # inspect x
x = torch.zeros(4, 8)  # 4x8 matrix of all zeros @inspect x
x = torch.ones(4, 8)  # 4x8 matrix of all ones @inspect x
x = torch.randn(4, 8)  # 4x8 matrix of iid Normal(0, 1) samples @inspect x
x.shape
x = torch.empty(4, 8)  # 4x8 matrix of uninitialized values @inspect x
nn.init.trunc_normal_(x, mean=0, std=1, a=-2, b=2)  # inspect x
x = torch.zeros(4, 8)  # @markdown x
assert x.dtype == torch.float32  # Default type
assert x.numel() == 4 * 8
assert x.element_size() == 4  # Float is 4 bytes
```

```python
assert get_memory_usage(x) == 4 * 8 * 4  # 128 bytes
#One matrix in the feedforward layer of GPT-3:
assert get_memory_usage(torch.empty(12288 * 4, 12288)) == 2304 * 1024 * 1024  # 2.3 GB
#...which is a lot!
```

```python
import torch

def same_storage(tensor1, tensor2):
  """Checks if two PyTorch tensors share the same underlying storage.

  Args:
    tensor1: The first PyTorch tensor.
    tensor2: The second PyTorch tensor.

  Returns:
    True if the tensors share the same storage, False otherwise.
  """
  return tensor1.storage().data_ptr() == tensor2.storage().data_ptr()
```

```python
#One can enforce a tensor to be contiguous first:
y = x.transpose(1, 0).contiguous().view(2, 3)  # inspect y
assert not same_storage(x, y)
```

```python
# prompt: device = get_device()
# implement get_device function

import torch

def get_device():
  """
  Gets the appropriate device (GPU if available, otherwise CPU).
  """
  if torch.cuda.is_available():
    return torch.device("cuda:0")  # Use the first available GPU
  else:
    return torch.device("cpu")
```

```python
if torch.cuda.is_available():
    B = 16384  # Number of points
    D = 32768  # Dimension
    K = 8192   # Number of outputs
else:
    B = 1024
    D = 256
    K = 64
device = get_device()
x = torch.ones(B, D, device=device)
w = torch.randn(D, K, device=device)
y = x @ w
```

```python
# We have one multiplication (x[i][j] * w[j][k]) and one addition per (i, j, k) triple.
actual_num_flops = 2 * B * D * K  # inspect actual_num_flops
```
